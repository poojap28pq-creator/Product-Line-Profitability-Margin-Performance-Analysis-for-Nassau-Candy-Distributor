```python
# Create Division Summary Table

division_summary = df.groupby('Division').agg({
    'Sales': 'sum',
    'Gross Profit': 'sum',
    'Units': 'sum'
}).reset_index()
```

```python
# Revenue Contribution %
division_summary['Revenue %'] = (
    division_summary['Sales'] /
    division_summary['Sales'].sum()
) * 100

# Profit Contribution %
division_summary['Gross Profit %'] = (
    division_summary['Gross Profit'] /
    division_summary['Gross Profit'].sum()
) * 100

# Round values
division_summary['Revenue %'] = (
    division_summary['Revenue %']
).round(2)

division_summary['Gross Profit %'] = (
    division_summary['Gross Profit %']
).round(2)

print(division_summary)
```

```python
plt.figure(figsize=(8,6))

plt.scatter(
    division_summary['Revenue %'],
    division_summary['Gross Profit %']
)

for i, txt in enumerate(division_summary['Division']):
    plt.annotate(
        txt,
        (
            division_summary['Revenue %'].iloc[i],
            division_summary['Gross Profit %'].iloc[i]
        )
    )

plt.xlabel("Revenue Contribution %")
plt.ylabel("Profit Contribution %")
plt.title("Division Profitability Comparison")

plt.show()
```

```python
# Aggregate Product Revenue & Profit

pareto_df = df.groupby('Product Name').agg({
    'Sales': 'sum',
    'Gross Profit': 'sum'
}).reset_index()

# Sort by Revenue
pareto_df = pareto_df.sort_values(
    by='Sales',
    ascending=False
)

# Revenue Contribution %
pareto_df['Revenue %'] = (
    pareto_df['Sales'] /
    pareto_df['Sales'].sum()
) * 100

# Cumulative Revenue %
pareto_df['Cumulative Revenue %'] = (
    pareto_df['Revenue %'].cumsum()
)

# Products Contributing to 80% Revenue
top_revenue_products = pareto_df[
    pareto_df['Cumulative Revenue %'] <= 80
]

print(top_revenue_products)
```

```python
num_revenue_products = len(top_revenue_products)
num_revenue_products
```

```python
# Calculate Percentage of Total Products

revenue_product_percentage = (
    num_revenue_products /
    len(pareto_df)
) * 100

print(revenue_product_percentage)
```

```python
plt.figure(figsize=(12,6))

plt.bar(
    pareto_df['Product Name'],
    pareto_df['Revenue %']
)

plt.plot(
    pareto_df['Product Name'],
    pareto_df['Cumulative Revenue %'],
    marker='o'
)

plt.axhline(80, color='red', linestyle='--', label='80% Threshold')

plt.legend()

plt.xticks(rotation=90)

plt.ylabel("Revenue Percentage")
plt.xlabel("Product Name")
plt.title("Pareto Analysis by Revenue")

plt.show()
```

```python
# Aggregate product profit
profit_pareto = df.groupby('Product Name')['Gross Profit'].sum().reset_index()

# Sort by profit descending
profit_pareto = profit_pareto.sort_values(
    by='Gross Profit',
    ascending=False
)

# Profit contribution %
profit_pareto['Gross Profit %'] = (
    profit_pareto['Gross Profit'] /
    profit_pareto['Gross Profit'].sum()
) * 100

# Cumulative profit %
profit_pareto['Cumulative Profit %'] = (
    profit_pareto['Gross Profit %'].cumsum()
)
```

```python
# Products Reaching 80% Profit
top_profit_products = profit_pareto[
    profit_pareto['Cumulative Profit %'] <= 80
]
```

```python
# Count Products
num_profit_products = len(top_profit_products)
num_profit_products
```

```python
# Calculate Percentage
profit_product_percentage = (
    num_profit_products /
    len(pareto_df)
) * 100

print(profit_product_percentage)
```

```python
# Plot
fig, ax1 = plt.subplots(figsize=(12,6))

# Bar chart
ax1.bar(
    profit_pareto['Product Name'],
    profit_pareto['Gross Profit %']
)

ax1.set_ylabel("Profit Contribution %")
plt.xticks(rotation=90)

# Line chart
ax2 = ax1.twinx()
plt.title("Pareto Analysis by Profit")
ax2.plot(
    profit_pareto['Product Name'],
    profit_pareto['Cumulative Profit %'],
    marker='o'
)

# 80% reference line
ax2.axhline(
    80,
    color='red',
    linestyle='--'
)

ax2.set_ylabel("Cumulative Profit %")
plt.show()
```

Pareto analysis reveals that a small subset of products contributes the majority of company profit, indicating strong profit concentration and dependency on flagship products. The business follows a classic Pareto distribution where a limited number of high-performing products drive overall financial performance.

```python
# Product-Level Cost Summary
cost_analysis = df.groupby('Product Name').agg({
    'Sales': 'sum',
    'Cost': 'sum',
    'Gross Profit': 'sum',
    'Units': 'sum'
}).reset_index()
```

```python
# Calculate Margin %

cost_analysis['Margin %'] = (
    cost_analysis['Gross Profit'] /
    cost_analysis['Sales']
) * 100
```

```python
# Calculate Cost-to-Sales Ratio

cost_analysis['Cost Ratio'] = (
    cost_analysis['Cost'] /
    cost_analysis['Sales']
)
```

```python
# Cost vs Sales Scatter Plot

# Create colors based on margin
colors = []

for margin in cost_analysis['Margin %']:
    
    if margin >= cost_analysis['Margin %'].median():
        colors.append('green')
        
    else:
        colors.append('red')
        
# Plot
plt.figure(figsize=(12,6))

plt.scatter(
    cost_analysis['Sales'],
    cost_analysis['Cost'],
    s=250,
    c=colors
)

# Add labels
for i, txt in enumerate(cost_analysis['Product Name']):
    
    plt.annotate(
        txt,
        (
            cost_analysis['Sales'].iloc[i],
            cost_analysis['Cost'].iloc[i]
        ),
        xytext=(10,-10),
        textcoords='offset points',
        fontsize=8
    )

plt.xlabel("Sales")

plt.ylabel("Cost")

plt.title("Cost vs Sales with Margin Diagnostics")

plt.tight_layout()

plt.show()
```

```python
# Identify Cost-Heavy, Margin-Poor Products

cost_heavy_products = cost_analysis[
    (cost_analysis['Cost Ratio'] > cost_analysis['Cost Ratio'].median()) &
    (cost_analysis['Margin %'] < cost_analysis['Margin %'].median())
]

print(cost_heavy_products)
```

```python
# Identify Pricing Inefficiencies

pricing_issues = cost_analysis[
    (cost_analysis['Sales'] > cost_analysis['Sales'].median()) &
    (cost_analysis['Margin %'] < cost_analysis['Margin %'].median())
]

print(pricing_issues)
```

```python
# Repricing Candidates
repricing_candidates = cost_analysis[
    cost_analysis['Margin %'] < 10
]

print(repricing_candidates)
```

```python
# Cost Renegotiation Candidates
cost_renegotiation = cost_analysis[
    cost_analysis['Cost Ratio'] > 0.8
]

print(cost_renegotiation)
```

```python
# Discontinuation Review Candidates

discontinuation_candidates = cost_analysis[
    (cost_analysis['Sales'] < cost_analysis['Sales'].median()) &
    (cost_analysis['Gross Profit'] < cost_analysis['Gross Profit'].median())
]

print(discontinuation_candidates)
```
